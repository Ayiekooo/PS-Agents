```python
import heapq

# Our graph from Part 1
graph = {
    'Nairobi': {'Nakuru': 161},
    'Nakuru': {'Kericho': 117, 'Eldoret': 156, 'Kisumu': 178},
    'Kericho': {'Kisumu': 84, 'Kisii': 104},
    'Kisii': {'Kisumu': 136},
    'Eldoret': {'Kisumu': 138, 'Kakamega': 106},
    'Kakamega': {'Kisumu': 81},
    'Kisumu': {}
}

def uniform_cost_search(start, goal):
    # Priority queue stores tuples
    pq = [(0, start, [start])]
    visited = []

    print("Starting UCS from", start, "to", goal, "\n")

    while pq:
        cost, node, path = heapq.heappop(pq)

        print("Visiting:", node, "| Cost:", cost, "| Path:", " -> ".join(path))

        if node == goal:
            print("\nGoal reached!")
            return cost, path

        if node not in visited:
            visited.append(node)

            for neighbor, distance in graph[node].items():
                if neighbor not in visited:
                    new_cost = cost + distance
                    new_path = path + [neighbor]
                    heapq.heappush(pq, (new_cost, neighbor, new_path))

    return float('inf'), []

# Run the search
start_city = 'Nairobi'
goal_city = 'Kisumu'

final_cost, final_path = uniform_cost_search(start_city, goal_city)


print("Best Path:", " -> ".join(final_path))
print("Total Cost:", final_cost, "km")
```
